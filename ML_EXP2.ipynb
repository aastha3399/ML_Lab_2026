```python
import pandas as pd
import numpy as np

data = pd.read_csv('breast-cancer.csv')

# 1. Handling Missing Values

# Fill numerical columns with mean
for col in data.select_dtypes(include=np.number).columns:
    data[col].fillna(data[col].mean(), inplace=True)

# Fill categorical columns with mode
for col in data.select_dtypes(include='object').columns:
    data[col].fillna(data[col].mode()[0], inplace=True)

print("\nMissing Values After Handling:")
print(data.isnull().sum())
```

```python
# 2. One-Hot Encoding

categorical_cols = data.select_dtypes(include='object').columns

data = pd.get_dummies(data, columns=categorical_cols)

print("\nDataset After One-Hot Encoding:")
print(data.head())
```

```python
# 3. Normalization (Min-Max Scaling)

# Normalize all columns except the last (Target)
X = data.iloc[:, :-1]

# Convert boolean columns in X to integer (0 or 1) before scaling
# This handles the TypeError caused by boolean subtraction
for col in X.select_dtypes(include='bool').columns:
    X[col] = X[col].astype(int)

X_normalized = (X - X.min()) / (X.max() - X.min())

# Add target column back
processed_data = X_normalized.copy()
processed_data[data.columns[-1]] = data.iloc[:, -1]

print("\nNormalized Dataset:")
print(processed_data.head())

# Save preprocessed dataset
processed_data.to_csv("preprocessed_dataset.csv", index=False)

print("\nPreprocessed dataset saved successfully.")
```
