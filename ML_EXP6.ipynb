**Decision Tree with Single Feature**

```python
import pandas as pd
import numpy as np
import math
```

```python
data = {
    'Outlook' : ['Sunny', 'Sunny', 'Rainy', 'Overcast', 'Rainy', 'Sunny', 'Overcast', 'Overcast', 'Rainy'],
    'Play' : ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
  }

df = pd.DataFrame(data)
print('Dataset : ')
print(df)
```

```python
def calculate_entropy(column):
  values = column.value_counts()
  print(values)
  total = len(column)
  print(total)
  entropy = 0
  for value in values:
    probability = value / total
    entropy -= probability * math.log2(probability)
  return entropy
```

```python
calculate_entropy(df['Play'])
```

```python
total_entropy = calculate_entropy(df['Play'])
print("Total Entropy : ", total_entropy)
print("Total Entropy(Round) : ", round(total_entropy, 3))
```

```python
def calculate_information_gain(df, feature, target):
  total_entropy = calculate_entropy(df[target])
  weighted_entropy = 0
  values = df[feature].unique()
  for value in values:
    subset = df[df[feature] == value]
    probability = len(subset) / len(df)
    subset_entropy = calculate_entropy(subset[target])
    weighted_entropy += probability * subset_entropy
    #print("Subset : ", subset)
    #print("Subset Entropy : ", subset_entropy)
  information_gain = total_entropy - weighted_entropy
  print("Weighted Entropy : ", weighted_entropy)
  print("Information Gain : ", information_gain)
  return information_gain
```

```python
calculate_information_gain(df, 'Outlook', 'Play')
```

```python
def entropy(data):
  values, counts = np.unique(data, return_counts=True)
  entropy = 0
  for count in counts:
    probability = count / len(data)
    entropy -= probability * np.log2(probability)
  return entropy
```
