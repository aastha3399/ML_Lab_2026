```python
#KNN (k-nearest neighbours)

import pandas as pd
import numpy as np
```

```python
# Load dataset
data = pd.read_csv("iris.csv")

# Input features
X = data[["sepal_length", "sepal_width",
          "petal_length", "petal_width"]].values

# Target
y = data["species"].values

# Split data: 80% training, 20% testing
np.random.seed(42)

indices = np.random.permutation(len(X))
train_size = int(0.8 * len(X))

train = indices[:train_size]
test = indices[train_size:]

X_train = X[train]
y_train = y[train]

X_test = X[test]
y_test = y[test]


# KNN function
def knn(X_train, y_train, test_point, k=5):

    # Calculate Euclidean distance
    distance = np.sqrt(np.sum((X_train - test_point) ** 2, axis=1))

    # Get nearest k points
    nearest = np.argsort(distance)[:k]

    # Get their classes
    classes = y_train[nearest]

    # Find majority class
    values, counts = np.unique(classes, return_counts=True)

    return values[np.argmax(counts)]


# Prediction
predictions = []

for point in X_test:
    predictions.append(knn(X_train, y_train, point, k=5))

# Accuracy
accuracy = np.mean(np.array(predictions) == y_test) * 100

print("KNN Classification")
print("-------------------")
print("K value:", 5)
print("Accuracy:", round(accuracy, 2), "%")

# Display first 10 predictions
print("\nActual       Predicted")

for i in range(10):
    print(y_test[i], "   ", predictions[i])
```
